```python
import pandas as pd
import numpy as np

df = pd.read_csv("11_adspend_vs_website_visits.csv")

df.head()
```

```python
df["ad_spend_usd"] = pd.to_numeric(
    df["ad_spend_usd"],
    errors="coerce"
)

df["marketing_spend_usd"] = pd.to_numeric(
    df["marketing_spend_usd"],
    errors="coerce"
)

df["competitor_price_usd"] = pd.to_numeric(
    df["competitor_price_usd"],
    errors="coerce"
)

df["season_index"] = pd.to_numeric(
    df["season_index"],
    errors="coerce"
)

df["website_visits"] = pd.to_numeric(
    df["website_visits"],
    errors="coerce"
)

df.dtypes
```
