```python
#2. Cyber-Force AI — Network Intrusion Risk Prediction

#Context: Network Intrusion Detection|Sensor records: 500
#Inputs: Packet_Size (Bytes); Request_Latency (ms); Encryption_Entropy (0-10)
#Target used for the runnable regression example: Threat_Risk_Score — predicted network threat-risk score (0–100).


# ANN Level 1 - FULL PROGRAM (Worksheet + Answers combined)
# 2. Cyber-Force AI — Network Intrusion Risk Prediction (Regression)


import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd
import numpy as np
import os
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
torch.manual_seed(42)
np.random.seed(42)
print("Libraries imported successfully! PyTorch Version:", torch.__version__)
```

```python
# [CELL 2] Data Pipeline, Normalization & Tensor Conversion

if not os.path.exists('cyber_force_network.csv'):
    n = 500
    demo_df = pd.DataFrame({
        'Packet_Size': np.random.randint(100, 3000, n),
        'Request_Latency': np.random.uniform(5, 200, n),
        'Encryption_Entropy': np.random.uniform(0, 10, n),
    })

    demo_df['Threat_Risk_Score'] = np.clip(
        5 + 0.012*demo_df['Packet_Size'] + 0.35*demo_df['Request_Latency']
        - 5.0*demo_df['Encryption_Entropy'] + np.random.normal(0, 5, n), 0, 100
    )
    for col in ['Packet_Size', 'Request_Latency']:
        mask = np.random.rand(n) < 0.03
        demo_df.loc[mask, col] = np.nan
    demo_df.to_csv('cyber_force_network.csv', index=False)

df = pd.read_csv('cyber_force_network.csv')

feature_cols = ['Packet_Size', 'Request_Latency', 'Encryption_Entropy']
X = df[feature_cols].values
y = df['Threat_Risk_Score'].values

imputer = SimpleImputer(strategy='mean')
X_imputed = imputer.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_imputed, y, test_size=0.2, random_state=42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_train_tensor = torch.tensor(X_train_scaled, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train, dtype=torch.float32).unsqueeze(1)
X_test_tensor = torch.tensor(X_test_scaled, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test, dtype=torch.float32).unsqueeze(1)
print(f"X_train shape: {X_train_tensor.shape}, y_train shape: {y_train_tensor.shape}")
```

```python
# [CELL 3] Designing the ANN Architecture for Regression

class ThreatRiskRegressor(nn.Module):
    def __init__(self):
        super(ThreatRiskRegressor, self).__init__()   
        self.linear1 = nn.Linear(3, 32)
        self.linear2 = nn.Linear(32, 16)
        self.linear3 = nn.Linear(16, 1)
        self.relu = nn.ReLU()
    def forward(self, x):
        x = self.relu(self.linear1(x))
        x = self.relu(self.linear2(x))
        x = self.linear3(x)
        return x
model = ThreatRiskRegressor()
print(model)
```

```python
# [CELL 4] Loss Function & Optimizer

loss_fn = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
print("Loss function: MSELoss | Optimizer: Adam")
```

```python
# [CELL 5] The Training Loop

num_epochs = 100
batch_size = 32
print("Training started...\n")
for epoch in range(num_epochs):
    for i in range(0, len(X_train_tensor), batch_size):
        X_batch = X_train_tensor[i:i+batch_size]
        y_batch = y_train_tensor[i:i+batch_size]
        predictions = model(X_batch)
        loss = loss_fn(predictions, y_batch)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    if (epoch + 1) % 20 == 0:
        print(f"Epoch [{epoch+1}/{num_epochs}], MSE Loss: {loss.item():.4f}")
print("\nTraining complete!")
```

```python
# [CELL 6] The Evaluation Loop

model.eval()
with torch.no_grad():
    test_predictions = model(X_test_tensor)
test_mse = loss_fn(test_predictions, y_test_tensor).item()
test_rmse = np.sqrt(test_mse)
print("Test Set Evaluation:")
print(f" - Mean Squared Error (MSE) : {test_mse:.4f}")
print(f" - Root Mean Squared Error (RMSE): {test_rmse:.4f}\n")
print(f"{'Actual':<15} {'Predicted':<15} {'Abs Error':<15}")
print("-" * 45)
with torch.no_grad():
    for i in range(5):
        actual = y_test_tensor[i].item()
        predicted = test_predictions[i].item()
        error = abs(actual - predicted)
        print(f"{actual:<15.2f} {predicted:<15.2f} {error:<15.2f}")
```

```python
# [CELL 7] Real-World Inference

raw_input = np.array([[1200, 85, 4.5]])
imputed_data = imputer.transform(raw_input)
scaled_data = scaler.transform(imputed_data)
input_tensor = torch.tensor(scaled_data, dtype=torch.float32)
model.eval()
with torch.no_grad():
    predicted_score = model(input_tensor).item()
print(f"Predicted Threat Risk Score: {predicted_score:.2f} / 100")
```
